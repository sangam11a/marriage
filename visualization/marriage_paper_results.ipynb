# Nepalese 21-Card Marriage AI: Q1 Journal Benchmark & Empirical Analysis

This notebook provides the complete empirical pipeline for the **21-Card Nepalese Marriage AI** benchmark targeting top-tier Q1 venues (*IEEE Transactions on Games*, *Artificial Intelligence*, *JAAMAS*).

### Methodological Highlights:
- **Partially Observable Stochastic Game (POSG)** with strictly verified zero-sum terminal payoffs ($\sum_{i=0}^3 \mathcal{R}_i = 0$).
- **Cyclic Duplicate Deals**: $5,120$ unique deal seeds rotated across 4 seats ($20,480$ total rounds), treating the deal seed as the paired unit of analysis.
- **Non-Parametric Statistical Inference**: $10,000$ paired bootstrap resamples, exact sign-flip permutation tests, and family-wise Holm-Bonferroni correction.
- **Capacity-Matched Representation Ablation**: Flat MLP vs. Permutation-Invariant DeepSets vs. Relational Graph Neural Network (R-GCN).
